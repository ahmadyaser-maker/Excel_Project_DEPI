```python
# =========================================================
# Detect Outliers Data:
# =========================================================

# ---------------------------------------------------------
# 1. List of Numeric Columns Selected:
# ---------------------------------------------------------
numeric_columns = [
    'air_temperature_K',
    'process_temperature_K',
    'rotational_speed_rpm',
    'vibration_mm_s',
    'pressure_bar',
    'voltage_V',
    'current_A',
    'torque_Nm',
    'tool_wear_min',
    'operating_hours',
    'maintenance_cost',
    'production_units',
    'defect_rate'
]

# ---------------------------------------------------------
# 2. Detect outliers and Print a Report:
# ---------------------------------------------------------
def report_outliers_iqr(dataframe, columns):
    outliers_count = {}
    
    for columns in columns:
        if columns in dataframe.columns:
            # Quarter Calculation:
            Q1 = dataframe[columns].quantile(0.25)
            Q3 = dataframe[columns].quantile(0.75)
            IQR = Q3 - Q1
            
            # Boundary Determination:
            lower_bound = Q1 - 1.5 * IQR
            upper_bound = Q3 + 1.5 * IQR
            
            # Calculating the Number of Outliers:
            outliers = dataframe[(dataframe[columns] < lower_bound) | (dataframe[columns] > upper_bound)]
            outliers_count[columns] = len(outliers)
            
            print(f"Column: {columns}")
            print(f"Lower Bound: {lower_bound}")
            print(f"Upper Bound: {upper_bound}")
            print(f"Number of Outliers Detected: {len(outliers)}\n")
        else:
            print(f"Warning: The column {columns} does not exist in the data.")
            
    return outliers_count

# ---------------------------------------------------------
# 3. Report Generation:
# ---------------------------------------------------------
outliers_summary = report_outliers_iqr(df_ai4i2020_fs, numeric_columns)
```

```python
# 1. List of Numeric Columns:
numeric_columns = [
    'air_temperature_K',
    'process_temperature_K',
    'temperature_C',
    'rotational_speed_rpm',
    'vibration_mm_s',
    'pressure_bar',
    'voltage_V', 'current_A',
    'torque_Nm',
    'tool_wear_min',
    'operating_hours',
    'maintenance_cost',
    'production_units',
    'defect_rate'
]

# 2. Plot Boxplots for all Columns:
def plot_outliers_boxplots(dataframe, columns):
    # Determine the number of rows and columns for the plot grid (e.g., 4 columns per row)
    number_columns = 4
    number_rows = (len(columns) + number_columns - 1) // number_columns
    
    # Create the Figure with Dimensions Suitable for the Number of Plots:
    figure, axes = plt.subplots(number_rows, number_columns, figsize=(20, 5 * number_rows))
    
    # Flatten the Axes Array to Make it Easier to Iterate Over:
    axes = axes.flatten()
    
    # Set the general theme for the plots
    sns.set_theme(style="whitegrid")
    
    for i, col in enumerate(columns):
        if col in dataframe.columns:
            # Plot the boxplot for the current column
            sns.boxplot(y=dataframe[col], ax=axes[i], color='skyblue', flierprops={'marker': 'o', 'markerfacecolor': 'red', 'markersize': 5})
            
            # Add and format the plot title
            axes[i].set_title(f'Outliers in {col}', fontsize=12, fontweight='bold')
            axes[i].set_ylabel('Values', fontsize=10)
        else:
            # If the column does not exist, hide the empty plot
            axes[i].set_visible(False)
            
    # Hide any remaining empty plots in the grid (if the number of columns does not fill the entire grid)
    for j in range(i + 1, len(axes)):
        axes[j].set_visible(False)
        
    # Adjust the plot layout to prevent overlapping
    plt.tight_layout()
    
    # Display the plots
    plt.show()

plot_outliers_boxplots(df_ai4i2020, numeric_columns)
```

```python
# =========================================================
# 1. MTTR - Mean Time To Repair
# Formula:
# SUM(downtime_hours) / SUM(machine_failure_flag)
# =========================================================
total_downtime_hours = df_ai4i2020_fs["downtime_hours"].sum()
total_failures = df_ai4i2020_fs["machine_failure"].sum()

mttr = total_downtime_hours / total_failures if total_failures > 0 else np.nan

# =========================================================
# 2. Cost/Failure
# Formula:
# SUM(maintenance_cost) / SUM(machine_failure)
# =========================================================
total_maintenance_cost = df_ai4i2020_fs["maintenance_cost"].sum()
total_failures = df_ai4i2020_fs["machine_failure"].sum()

cost_per_failure = total_maintenance_cost / total_failures if total_failures > 0 else np.nan

# =========================================================
# 3. Good Units
# Formula:
# production_units * (1 - defect_rate_decimal)
# =========================================================
df_ai4i2020_fs["good_units"] = (df_ai4i2020_fs["production_units"] * (1 - df_ai4i2020_fs["defect_rate"]))

total_good_units = df_ai4i2020_fs["good_units"].sum()

# =========================================================
# 4. Operating Hours
# Formula:
# SUM(operating_hours)
# =========================================================
total_operating_hours = df_ai4i2020_fs['operating_hours'].sum()

# =========================================================
# 5. Maintenace Cost
# Formula:
# SUM(maintenace_cost)
# =========================================================
total_maintenance_cost = df_ai4i2020_fs['maintenance_cost'].sum()


# =========================================================
# Display KPI Results
# =========================================================

kpi_results = pd.DataFrame({
    "KPI": [
        "MTTR",
        "Cost per Failure",
        "Good Units",
        "Operating Hours",
        "Maintenace Cost"
    ],

    "Value": [
        f"{mttr:.0f} h",
        f"{cost_per_failure / 1000:.1f} K",
        f"{total_good_units / 1_000_000:.1f} M",
        f"{total_operating_hours / 1000:.1f} K",
        f"{total_maintenance_cost / 1_000_000:.1f} M"
    ]
})

# Display Final KPI Table
kpi_results
```

```python
# Calculate Sum of Maintenance Cost by Failure Type
failure_cost_analysis = (
    df_ai4i2020_fs
    .groupby("failure_type", as_index=False)["maintenance_cost"]
    .sum()
)

# Set the exact order shown in the Excel Pivot Table
failure_order = [
    "None",
    "Overheat",
    "Overstrain",
    "Power",
    "Unknown",
    "Wear"
]

failure_cost_analysis["failure_type"] = pd.Categorical(
    failure_cost_analysis["failure_type"],
    categories=failure_order,
    ordered=True
)

# Sort by the required order
failure_cost_analysis = (
    failure_cost_analysis
    .sort_values("failure_type")
    .reset_index(drop=True)
)

# Calculate Grand Total
grand_total = failure_cost_analysis["maintenance_cost"].sum()

# Function to format values as K or M
def format_cost(value):
    if value >= 1_000_000:
        return f"{value / 1_000_000:.1f} M"
    elif value >= 1_000:
        return f"{value / 1_000:.1f} K"
    return f"{value:.1f}"

# Create formatted cost column
failure_cost_analysis["Formatted Cost"] = (
    failure_cost_analysis["maintenance_cost"]
    .apply(format_cost)
)

# Create final result
result = failure_cost_analysis[
    ["failure_type", "Formatted Cost"]
].rename(
    columns={
        "failure_type": "Failure Type",
        "Formatted Cost": "Sum of Maintenance Cost"
    }
)

# Display result
result
```
